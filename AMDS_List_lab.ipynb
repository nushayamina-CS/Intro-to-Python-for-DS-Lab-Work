```python
33# P2: Take ten integers from the user and store them using append()

numbers = []

print("Enter 10 integers:")

for i in range(10):
    num = int(input())
    numbers.append(num)

print("Complete List:", numbers)
print("Total Elements:", len(numbers))
```

```python
# P3: Calculate sum, average, maximum and minimum without built-in functions

numbers = list(map(int, input("Enter integers separated by space: ").split()))

total = 0
maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    total += num

    if num > maximum:
        maximum = num

    if num < minimum:
        minimum = num

average = total / len(numbers)

print("Sum =", total)
print("Average =", average)
print("Maximum =", maximum)
print("Minimum =", minimum)
```

```python
# P4: Count even and odd numbers

numbers = list(map(int, input("Enter integers: ").split()))

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even Elements =", even)
print("Odd Elements =", odd)
```

```python
# P5: Create a new list containing only even numbers

numbers = list(map(int, input("Enter integers: ").split()))

even_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)

print("Even Numbers:", even_numbers)
```

```python
# P6: Search an element in a list and display its index

numbers = list(map(int, input("Enter integers: ").split()))
target = int(input("Enter number to search: "))

found = False

for i in range(len(numbers)):
    if numbers[i] == target:
        print("Number found at index", i)
        found = True
        break

if not found:
    print("Number not found.")
```

```python
# P7: Count frequency of each element

numbers = list(map(int, input("Enter integers: ").split()))

visited = []

for num in numbers:
    if num not in visited:
        count = 0
        for item in numbers:
            if item == num:
                count += 1
        print(num, "->", count)
        visited.append(num)
```

```python
# P8: Remove duplicate elements

numbers = list(map(int, input("Enter integers: ").split()))

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("List without duplicates:", unique)
```

```python
# P15
numbers = list(map(int, input().split()))

total = 0
largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    total += num

    if num > largest:
        largest = num

    if num < smallest:
        smallest = num

average = total / len(numbers)

print("Sum:", total)
print("Average:", average)
print("Largest:", largest)
print("Smallest:", smallest)
```

```python
#P17

matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print("Matrix:")
for row in matrix:
    print(row)

print("First Row:", matrix[0])

print("Last Column:")
for row in matrix:
    print(row[-1])

total = 0

for row in matrix:
    for value in row:
        total += value

print("Sum:", total)
```

```python
# P18

matrix = []

print("Enter 3x3 matrix:")

for i in range(3):
    row = list(map(int, input().split()))
    matrix.append(row)

print("Row Sums:")

for row in matrix:
    print(sum(row))

print("Column Sums:")

for col in range(3):
    total = 0
    for row in range(3):
        total += matrix[row][col]
    print(total)

total = 0

for row in matrix:
    for value in row:
        total += value

print("Total Sum:", total)
```

```python
# P19

marks = []

print("Enter marks of 10 students:")

for i in range(10):
    marks.append(float(input()))

highest = marks[0]
lowest = marks[0]
total = 0

for mark in marks:
    total += mark

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / len(marks)

count = 0

for mark in marks:
    if mark > average:
        count += 1

print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
print("Above Average:", count)
```

```python
# P20

runs = []

print("Enter runs of 15 matches:")

for i in range(15):
    runs.append(int(input()))

highest = runs[0]
lowest = runs[0]
total = 0

half = 0
century = 0

for score in runs:
    total += score

    if score > highest:
        highest = score

    if score < lowest:
        lowest = score

    if score >= 50:
        half += 1

    if score >= 100:
        century += 1

average = total / len(runs)

print("Total Runs:", total)
print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
print("Half-centuries:", half)
print("Centuries:", century)
```
